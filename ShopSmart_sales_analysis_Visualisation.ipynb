```python
import matplotlib.pyplot as plt
import pandas as pd

cities = ["Lagos", "Abuja", "Port Harcourt", "Ibadan"]
revenue = [750000, 175000, 50000, 25000]

plt.figure(figsize=(10, 6))
plt.bar(cities, revenue, color=["blue", "green", "red", "orange"])
plt.title("TradeMart Revenue by City")
plt.xlabel("City")
plt.ylabel("Revenue (₦)")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [45000, 52000, 48000, 61000, 55000, 70000]

plt.figure(figsize=(10, 6))
plt.plot(months, sales, color="blue", marker="o", linewidth=2)
plt.title("Monthly Sales Trend 2026")
plt.xlabel("Month")
plt.ylabel("Sales (₦)")
plt.grid(True)
plt.show()
```

```python
cities = ["Lagos", "Abuja", "Port Harcourt", "Ibadan"]
revenue = [750000, 175000, 50000, 25000]

plt.figure(figsize=(8, 8))
plt.pie(revenue, labels=cities, autopct="%1.1f%%",
        colors=["blue", "green", "red", "orange"])
plt.title("Revenue Distribution by City")
plt.show()
```

```python
age = [32, 28, 35, 25, 40, 29, 33]
salary = [150000, 75000, 50000, 150000, 25000, 100000, 450000]

plt.figure(figsize=(10, 6))
plt.scatter(age, salary, color="purple", s=150)
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary (₦)")
plt.grid(True)
plt.show()
```

```python
ages = [32, 28, 35, 25, 40, 29, 33, 28, 32, 35, 29, 40, 25, 33, 28]

plt.figure(figsize=(10, 6))
plt.hist(ages, bins=5, color="teal", edgecolor="black")
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
import seaborn as sns

data = {
    "Product": ["Laptop", "Phone", "Headphones", "Tablet"],
    "Revenue": [900000, 450000, 225000, 150000]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))
sns.barplot(x="Product", y="Revenue", data=df, palette="Blues_d")
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₦)")
plt.show()
```

```python
import seaborn as sns

data = {
    "Product": ["Laptop", "Phone", "Headphones", "Tablet"],
    "Revenue": [900000, 450000, 225000, 150000]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))
sns.barplot(x="Product", y="Revenue",
            data=df, palette="Blues_d",
            hue="Product", legend=False)
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₦)")
plt.show()
```
