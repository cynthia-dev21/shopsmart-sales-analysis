```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Chidi", "amaka", "Chidi", "EMEKA", "ngozi", "Tunde", "kemi"],
    "City": ["Lagos", "abuja", "Lagos", None, "Lagos", "Ibadan", "ABUJA"],
    "Age": [32, 28, 32, 35, 200, None, 29],
    "Salary": [150000, 75000, 150000, 50000, 38000, None, 100000]
}

df = pd.DataFrame(data)
print("Original dirty data:")
print(df)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("Duplicates:", df.duplicated().sum())

print("Data types:")
print(df.dtypes)
```

```python
df = df.drop_duplicates()
print("After removing duplicates:")
print(df)
print("Shape:", df.shape)
```

```python
# Fill missing Age with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing City with Unknown
df["City"] = df["City"].fillna("Unknown")

# Fill missing Salary with 0
df["Salary"] = df["Salary"].fillna(0)

print("After handling missing values:")
print(df)
```

```python
# Fix name and city capitalisation
df["Name"] = df["Name"].str.title()
df["City"] = df["City"].str.title()

print("After fixing text:")
print(df)
```

```python
# Remove impossible ages
df = df[df["Age"] <= 100]

print("After removing impossible ages:")
print(df)
print("Final shape:", df.shape)
```

```python
# Convert Age to integer
df["Age"] = df["Age"].astype(int)

print("Final clean dataset:")
print(df)

print("Any missing values left?", df.isnull().sum().any())
```
